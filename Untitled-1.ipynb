# Missing Values 

```python
df["photos_count"] = df["photos_count"].fillna(0)
```

```python
df["transmission"] = df["transmission"].fillna(df["transmission"].mode()[0])
df["color"] = df["color"].fillna(df["color"].mode()[0])
df["mileage_km"] = df["mileage_km"].fillna(df["mileage_km"].median())
df["fuel_type"] = df["fuel_type"].fillna(df["fuel_type"].mode()[0])
df["engine_cc"] = df["engine_cc"].fillna(df["engine_cc"].median())
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["seller_type"] = df["seller_type"].fillna(df["seller_type"].mode()[0])
df["previous_owners"] = df["previous_owners"].fillna(df["previous_owners"].median())
df["condition_rating"] = df["condition_rating"].fillna(df["condition_rating"].mode()[0])


```

```python
numeric_cols = [
    "year",
    "mileage_km",
    "engine_cc",
    "previous_owners",
    "condition_rating",
    "photos_count",
    "price_egp"
]

for col in numeric_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()
```

```python
for col in ["price_egp", "mileage_km", "year"]:
    plt.figure(figsize=(7, 4))
    sns.boxplot(
        data=df,
        x="sold_within_30_days",
        y=col
    )
    plt.title(f"{col} vs Sold Within 30 Days")
    plt.tight_layout()
    plt.show()
```

```python

categorical_cols = [
    "brand",
    "transmission",
    "fuel_type",
    "body_type",
    "color",
    "city",
    "seller_type",
    "accident_history"
]

for col in categorical_cols:
    rates = (
        df.groupby(col, dropna=False)["sold_within_30_days"]
        .apply(lambda s: s.astype(str).str.strip().str.lower()
               .isin(["yes", "y", "1", "true"]).mean() * 100)
        .sort_values(ascending=False)
    )

    print(f"\n{col} - Sale Rate (%):")
    print(rates)

    rates.plot(kind="bar", figsize=(9, 4))
    plt.title(f"Sale Rate by {col}")
    plt.ylabel("Sold Within 30 Days (%)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
# Get the top 3 most frequent cities
top_3_cities = df["city"].value_counts().nlargest(3).index

# Replace all other cities with "Others"
df["city"] = df["city"].where(
    df["city"].isin(top_3_cities),
    "Others"
)

# Check the result
print(df["city"].value_counts())
```

```python
categorical_cols = [
    "brand_model",
    "fuel_type",
    "color",
    "city",
]

ohe = OneHotEncoder(
    drop="first",
    handle_unknown="ignore",
    sparse_output=False
)

encoded_data = ohe.fit_transform(df[categorical_cols])

encoded_df = pd.DataFrame(
    encoded_data,
    columns=ohe.get_feature_names_out(categorical_cols),
    index=df.index
)

df = pd.concat(
    [df.drop(columns=categorical_cols), encoded_df],
    axis=1
)
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[["no", "yes"]]
)

cols = ["sold_within_30_days", "accident_history"]

for col in cols:
    df[[col]] = encoder.fit_transform(df[[col]])
```
